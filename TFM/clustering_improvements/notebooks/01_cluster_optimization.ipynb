# 📊 Optimización del Número de Clusters

**Objetivo**: Determinar el número óptimo de clusters usando múltiples métricas

## Métricas a evaluar:
- **Silhouette Score**: Mide qué tan similar es un objeto a su cluster vs otros clusters (rango: -1 a 1, mayor es mejor)
- **Davies-Bouldin Index**: Ratio de distancia intra-cluster vs inter-cluster (menor es mejor)
- **Calinski-Harabasz Score**: Ratio de dispersión entre clusters vs dentro clusters (mayor es mejor)
- **Inertia (SSE)**: Suma de distancias al cuadrado (método del codo)

## Configuración actual:
- Clusters originales: **6**
- Algoritmo: K-Means con PCA
